```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col,udf
from pyspark.sql.types import StringType, IntegerType
import time
```

```python
spark = SparkSession.builder.appName("SparkCompleteNotes").getOrCreate()
# Create Base DataFrame
data = [
    ("Alice", 16),
    ("Bob", 22),
    ("Charlie", 17),
    ("David", 28),
    ("Eve", 15),
    ("Frank", 29),
    ("Grace", 18),
    ("Hannah", 14),
    ("Ian", 40),
    ("Jack", 13),
    ("Karen", 34),
    ("Leo", 19),
    ("Mia", 12),
    ("Nathan", 36),
    ("Olivia", 17),
    ("Paul", 41),
    ("Quinn", 20),
    ("Rachel", 16),
    ("Steve", 35),
    ("Tina", 21)
]

columns = ["Name", "Age"]

df = spark.createDataFrame(data, columns)
```

```python
def cat_age(age):
    if age>=18:
        return "can drink"
    return "can't drink"
```

```python
age_cat_udf=udf(cat_age, StringType())
res=df.withColumn("Category", age_cat_udf(col("Age")))
print("Method !: Standard udf")
res.show()
```

```python
spark.udf.register("sql_cat_age", cat_age, StringType())
df.createOrReplaceTempView("people")
```

```python
sql_df=spark.sql("""
    select
        name,
        age,
        sql_cat_age(Age) as category
        from people
""")
sql_df.show()
```
